# Load in jalview extracted IDR fasta file

```python
mainprotein='catsper1_n_terminals'

## Load in sequences
myfile = open(mainprotein+'.fasta', 'r')
Lines = myfile.readlines()

tmpaccs=[]
tmpseqs=[]
thisseq = ''
for line in Lines:
    
    # remove pesky newline
    cleanline = line.strip()
    
    
    if cleanline[0]=='>':
        tmpaccs.append(cleanline)
        
        # check that this is not the very first sequence, if not 
        # add the sequence
        if (thisseq is not ''): 
            tmpseqs.append(thisseq.upper()) 

            # reset the sequence for the coming entry
            thisseq = '' 

    # else append the sequence
    else:
        thisseq += cleanline

# store the final sequence	
tmpseqs.append(thisseq.upper()) 

print(len(tmpaccs))
print(len(tmpseqs))

myseqs=[]
accs=[]
for s in range(0,len(tmpseqs)): 
    currseq=tmpseqs[s].strip('-')
    #print(currseq)
    #print(len(currseq))
    if len(currseq)>=30 and "X" not in currseq and "U" not in currseq and "Z" not in currseq and "J" not in currseq and "B" not in currseq and "O" not in currseq and "*" not in currseq:
        myseqs.append(currseq)
        accs.append(tmpaccs[s])
    else:
        print('BAD SEQUENCE: '+tmpaccs[s])
        
            
print(len(accs))
print(len(myseqs))

#print(accs)


with open(mainprotein+'_remove_bad.fasta', 'w') as myfile:
    for s in range(0,len(myseqs)):
        myfile.write(accs[s]+'\n')
        myfile.write(myseqs[s]+'\n')
        
with open(mainprotein+'_remove_bad_seqlist.txt','w') as myfile2:
    for s in range(0,len(myseqs)):
        myfile2.write(myseqs[s]+'\n')
```
